```python
import pandas as pd
```

```python
data = {
    'Player':['Virat Kohali', 'Rohit Sharma', 'Suryakumar Yadav','MS Dhoni', 'K L Rahul', 'Shubhaman Gill', 'Hardik Pandya', 'Ravindra Jadeja', 'Bhuvneshwar Kumar', 'Jasprit Bumrah', 'Mohammad Shami'],
    'Runs':[100, 67, 24, 36, 50, 27, 68, 15, 10, 5, 0],
    'Wickets':[0, 0, 0, 0, 1, 0, 2, 3, 1, 2, 4],
    'Balls Faced': [47, 30, 7, 20, 15, 6, 30, 8, 2, 4, 1],
    'Balls Bowled': [0, 0, 0, 0,  0, 4, 0, 10, 10, 10, 10]
}
match_data = pd.DataFrame(data);
```

```python
total_runs = match_data['Runs'].sum()
print("Total Runs:", total_runs)
```

```python
match_data['Batting Averages'] = match_data['Runs'] / (match_data['Balls Faced'] - match_data['Runs'])
print("\n Batting Averages:\n", match_data[['Player', 'Batting Averages']])
```

```python
bowlers =  match_data[match_data['Balls Bowled']>0]
bowlers['Bowling Strike Rate'] = bowlers['Balls Bowled'] / bowlers['Wickets']
print("\n Bowling Strike Rates:\n", bowlers[['Player', 'Bowling Strike Rate']])
```

```python
top_scorer = match_data.loc[match_data['Runs'].idxmax(), 'Player']
print("\nTop Run-Scorer:", top_scorer)
```

```python
top_wicket_taker = bowlers.loc[bowlers['Wickets'].idxmax(), 'Player']
print("Top Wicket-taker:", top_wicket_taker)
```

```python
match_data['Points'] = (match_data['Runs'] // 10) + (match_data['Wickets'] * 20)
```

```python
print("\nPlayer Points:\n", match_data[['Player', 'Points']])
```
